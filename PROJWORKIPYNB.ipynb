```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Read the file
df =pd.read_csv("PR2.csv")
```

```python
#information of the data.
df.info()
```

```python
#finding the sum of null values.
df.isnull().sum().sum()
```

```python
#finding how many null values in which particular columns.
df.isnull().sum()
```

```python
#Fetch the top 5 rows of the data.
df.head()
```

```python
#Drop the columns which is not necessary.
df.drop("reviews.id", axis =1 , inplace = True)
```

```python
df.drop("id",axis =1 ,inplace = True)
df.drop("manufacturerNumber", axis =1 , inplace = True)
df.drop("asins", axis =1 , inplace = True)
df.drop("keys", axis =1 , inplace = True)
df.drop("reviews.dateSeen", axis =1 , inplace = True)
df.drop("dateAdded", axis =1 , inplace = True)
df.drop("dateUpdated", axis =1 , inplace = True)
df.drop("reviews.didPurchase", axis =1 , inplace = True)
df.drop("reviews.sourceURLs", axis =1 , inplace = True)
df.drop("sourceURLs", axis =1 , inplace = True)
```

```python
df.columns.tolist()
```

```python
#RENAME THE COLUMNS NAME.
df.rename(columns={"reviews.date":"date"},inplace = True)
df.rename(columns={"reviews.doRecommend":"recommendation"},inplace = True)
df.rename(columns={"reviews.numHelpful":"numhelpful"},inplace = True)
df.rename(columns={"reviews.rating":"rating"},inplace = True)
df.rename(columns={"reviews.text":"text"},inplace = True)
df.rename(columns={"reviews.title":"title"},inplace = True)
df.rename(columns={"reviews.username":"username"},inplace = True)
```

```python
df.columns.tolist()
```

```python
df.isnull().sum()
```

```python
#Checking for the outliers.
sns.boxplot(x=df["numhelpful"])
plt.show()
#yes , here outliers is present.
#most values: 0–5
#extreme values: 100, 300, 600+
```

```python
#Outliers via IQR methods.
#STEP :1
Q1 = df['numhelpful'].quantile(0.25)
Q3 = df['numhelpful'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
```

```python
#STEP :2 CAP VALUES --->> HERE EXTREME VLAUES IS NOW REDUCED NOT REMOVE..
df['numhelpful'] = df['numhelpful'].clip(lower, upper)
```

```python
#STEP:3 REMOVE OUTLIERS -->> ONLY IF DATA IS CORRUPTED.
df = df[
    (df['numhelpful'] >= lower) &
    (df['numhelpful'] <= upper)
]
```

```python
df.isnull().sum()
```

```python
#Checking columns which existed.
df.columns.tolist()
```

```python
# Normal numeric data -->> Mean
# Numeric data with outliers -->>> Median
# Categorical/text data	->>> Mode
```

```python
#filling null values.
df["recommendation"] = df["recommendation"].fillna(False)
df["username"] =df["username"].fillna("Unknown")
```

```python
#BASICS OF FEATURE ENGINEERING 
#SENTIMENT COLUMNS:sentiment column means a new column you create to show the emotional meaning of a review based on its rating or text.
#CONVERTS DATA INTO THE MORE HUMAN FRIENDLY LANG LIKE POSTIVIVE , NEGATIVE , NEUTRAL.
def sentiment(x):
    if x >= 4:
        return "Positive"
    elif x == 3:
        return "Neutral"
    else:
        return "Negative"

df['sentiment'] = df['rating'].apply(sentiment)
print(df['sentiment'])
```

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine


# =========================================================
# 2. LOADING AND CLEANING DATA
# =========================================================

# Load dataset
df = pd.read_csv("amazon.csv")

# Display first 5 rows
df.head()

# Check missing values
df.isnull().sum()

# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Convert rating column into numeric
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

# Remove null ratings
df = df.dropna(subset=['rating'])

print("Data Cleaning Completed")


# =========================================================
# 3. DATA VISUALIZATION
# =========================================================

# Rating Distribution

plt.figure(figsize=(8,5))

sns.countplot(x='rating', data=df)

plt.ylabel("Count")

plt.show()

# =========================================================

password = "your_password"
host = "localhost"
database = "amazon"

    f"mysql+pymysql://{username}:{password}@{host}/{database}"
)

print("MySQL Connection Successful")

# 5. EXPORTING DATA TO MYSQL & SQL ANALYSIS
# =========================================================

# Export dataframe to MySQL

df.to_sql(

# SQL Query

query = """

SELECT category,
AVG(rating) AS average_rating
FROM amazon_reviews
GROUP BY category
ORDER BY average_rating DESC
LIMIT 10;

"""
